#### ii) *Where The Restaurants Are Located*

```python
# Using the coordinates to find the median location of each country, and the most common cities in each country. This will help identify the countries that are in the dataset.
real_loc = df[(df['Longitude'] != 0) | (df['Latitude'] != 0)]

evidence = real_loc.groupby('Country Code').agg(
    restaurants=('Restaurant ID', 'count'),
    example_cities=('City', lambda s: ', '.join(s.value_counts().index[:3])),   # 3 most common cities
    currency_labels=('Currency', lambda s: ', '.join(s.unique())),
    median_lat=('Latitude', 'median'),
    median_lon=('Longitude', 'median'),
).round(1)

evidence
# there are 15 countries in the dataset
```

```python
# Converting country codes to country names
country_map = {1: "India", 14: "Australia", 30: "Brazil", 37: "Canada",
           94: "Indonesia", 148: "New Zealand", 162: "Philippines",
           166: "Qatar", 184: "Singapore", 189: "South Africa",
           191: "Sri Lanka", 208: "Turkey", 214: "UAE",
           215: "United Kingdom", 216: "United States"}
country = df['Country Code'].map(country_map)
print('Codes with no name:', country.isna().sum()) # check if there are any country codes withut names in the map.
```

```python
# check the number of restaurants in each country
country_count= country.value_counts()
country_count

# India has the highest number of restaurants in the dataset, followed by the United States and the United Kingdom.
# Canada has the least number of restaurants
```

```python
#charting the countries using a bar chart
country_count[::-1].plot(kind='barh') # [::-1] reverses the order so the biggest bar ends up at the top.
plt.title('Number of restaurants per country')
plt.xlabel('Restaurants')
plt.show()
```

```python
# checking for cities with the most restaurants in the dataset
city_count = df['City'].value_counts().head(10)
city_count

# New Delhi has the highest number of datasets followed by Gurgaon and Noida
#
```

```python
# Chart the top 10 cities
city_count[::-1].plot(kind='barh')
plt.title('Top 10 cities by number of restaurants')
plt.xlabel('Restaurants')
plt.show()
```

```python
# checking the neighborhoods have the most restaurants
# sorting by both city and locality
neighborhood_counts = (
    df.groupby(['City', 'Locality'])
      .size()
      .sort_values(ascending=False)
      .head(10)
)
neighborhood_counts
# New delhi (ND) connaught place has the highest number of restaurants, followed by ND Rajouri Garden and ND Shahdara 
```

```python
# mapping the restaurants
# Some rows have Longitude = 0 AND Latitude = 0 (missing location). These are placeholder values, not real locations. We can filter them out when plotting the map. The ~ symbol means "not".
has_coords = ~((df['Longitude'] == 0) & (df['Latitude'] == 0))
print('Restaurants with placeholder (0, 0) location:', (~has_coords).sum())

# Drawing one dot per restaurant: x = longitude, y = latitude.
# s = dot size, alpha = transparency (so overlapping dots look darker).
plt.figure(figsize=(12, 6))
plt.scatter(df.loc[has_coords, 'Longitude'], df.loc[has_coords, 'Latitude'], s=4, alpha=0.4)
plt.title('All restaurants (longitude vs latitude)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
# focus on Delhi since most of the data is centred in the Delhi capital region, with a few outliers in other parts of India and the world.
# .between(a, b) is True when a value is inside the range. We combine conditions with &.
# These ranges roughly box in Delhi, Gurgaon, Noida, Faridabad and Ghaziabad.
delhi_capital = has_coords & df['Longitude'].between(76.8, 77.7) & df['Latitude'].between(28.2, 28.9)

plt.figure(figsize=(8, 8))
plt.scatter(df.loc[delhi_capital, 'Longitude'], df.loc[delhi_capital, 'Latitude'], s=5, alpha=0.3)
plt.title('Restaurants in Delhi Capital region')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

#### iii) *The Restaurant Price Ranges*

```python
#the price range of the restraurant in the dataset is from 1 to 4, with 1 being the cheapest and 4 being the most expensive.
# using a minimum restaurant count of 20
min_restaurant_count = 20

# NOTE: Outside the Delhi capital region, most cities in this dataset have exactly 20 restaurants each

# Count restaurants per city, then keep the names of cities that meet the minimum
city_count = df['City'].value_counts()
big_cities = city_count[city_count >= min_restaurant_count].index

# A filtered view of df containing only restaurants from those cities (df itself is unchanged)
df_big = df[df['City'].isin(big_cities)]
print('Cities kept:', len(big_cities), 'out of', df['City'].nunique())
print('Restaurants kept:', len(df_big), 'out of', len(df))
```

```python
# checking how restaurants are distributed across price ranges in the dataset
# sort_index() puts the rows in order 1, 2, 3, 4 instead of by size.
price_counts = df['Price range'].value_counts().sort_index()
print(price_counts)
```

```python
# charting the price range distribution
price_counts.plot(kind='bar')
plt.title('Restaurants by price range (1 = cheapest, 4 = most expensive)')
plt.xlabel('Price range')
plt.ylabel('Restaurants')
plt.xticks(rotation=0)
plt.show()
```

```python
# cities with the most expensive restaurants (price range 4)
expensive_cities = (df_big['Price range'] == 4).groupby(df_big['City']).mean() * 100

expensive_cities.sort_values(ascending=False).head(10).round(1)

# doha, pretoria, singapore and Sao paulo have the highest percentage of expensive restaurants in the dataset.
```

```python
# countries where the most expensive restaurants are located (price range 4)
country_price = pd.DataFrame({
    'restaurants': country.value_counts(),
    'pct_range_4': ((df['Price range'] == 4).groupby(country).mean() * 100).round(1),
})
country_price.sort_values('pct_range_4', ascending=False)

# Qatar, Singapore, South Africa and Brazil have the highest percentage of expensive restaurants in the dataset.
```

#### iv) *Restaurants By services: online delivery & table bookings*

```python
# % of restaurants that offer table booking and online delivery
# value_counts(normalize=True) gives shares (0 to 1) instead of counts; times 100 = percent.
print('Online Delivery booking (%):')
print((df['Has Online delivery'].value_counts(normalize=True) * 100).round(1))
print()
print('Table booking (%):')
print((df['Has Table booking'].value_counts(normalize=True) * 100).round(1))
```

```python
# Cities with the highest booking and delivery coverage
# (column == 'Yes') then average it per city.
coverage = pd.DataFrame({
    'restaurants': df_big.groupby('City').size(),
    'table_booking_%': ((df_big['Has Table booking'] == 'Yes').groupby(df_big['City']).mean() * 100).round(1),
    'online_delivery_%': ((df_big['Has Online delivery'] == 'Yes').groupby(df_big['City']).mean() * 100).round(1),
})

print('Top 10 cities for TABLE BOOKING coverage:')
display(coverage.sort_values('table_booking_%', ascending=False).head(10))

print('Top 10 cities for ONLINE DELIVERY coverage:')
display(coverage.sort_values('online_delivery_%', ascending=False).head(10))
```

```python
# HUsing a crosstab to see how many restaurants offer both services, one, or neither?
# Rows = table booking, columns = online delivery. margins=True adds the totals.
pd.crosstab(df['Has Table booking'], df['Has Online delivery'], margins=True)
```

```python
# Are pricier restaurants more likely to offer table booking?
# Rows = price range, columns = Has Table booking.
# normalize='index' turns each ROW into percentages that add up to 100%.
booking_by_price = (pd.crosstab(df['Price range'], df['Has Table booking'], normalize='index') * 100).round(1)
print(booking_by_price)
```

```python
# Charting the 'Yes' column
booking_by_price['Yes'].plot(kind='bar')
plt.title('% of restaurants with table booking, by price range')
plt.xlabel('Price range')
plt.ylabel('% with table booking')
plt.xticks(rotation=0)
plt.show()
```

```python
# Are cheaper restaurants more likely to offer online delivery?
# focus on the delivery column.
delivery_by_price = (pd.crosstab(df['Price range'], df['Has Online delivery'], normalize='index') * 100).round(1)
print(delivery_by_price)
```

```python
# charting the yes column
delivery_by_price['Yes'].plot(kind='bar', color='tab:brown')
plt.title('% of restaurants with online delivery, by price range')
plt.xlabel('Price range')
plt.ylabel('% with online delivery')
plt.xticks(rotation=0)
plt.show()
```

#### vi) *Restaurant names By popularity*

```python
# most popular restaurant names
name_counts = df['Restaurant Name'].value_counts()
name_counts.head(15)
```

```python
# Count of unique restaurant names vs total outlets?
total_outlets = len(df)
unique_names = df['Restaurant Name'].nunique()   # nunique() = number of different values
print('Total rows (outlets):', total_outlets)
print('Unique restaurant names:', unique_names)
print('Average outlets per name:', round(total_outlets / unique_names, 2))
```

```python
# Names that appear only once vs names that appear 2+ times (likely restaurant chains)
print('Names appearing only once:', (name_counts == 1).sum())
print('Names appearing 2+ times:', (name_counts >= 2).sum())
```

```python
# % of ALL outlets that belong to a name that appears 2+ times?
chain_outlets = name_counts[name_counts >= 2].sum()
print('% of outlets that belong to repeated names (%):', round(chain_outlets / total_outlets * 100, 1))
```

#### vii) *Analysing the cost in India as its the most popular country*

```python
# Keep only rows where Country Code is 1 (India)
df_india = df[df['Country Code'] == 1]
print('Indian restaurants:', len(df_india))
```

```python
# checking that they all use the same currency
print(df_india['Currency'].unique())
```

```python
# The average cost for two in India?
# Mean = the average. Median = the middle value (less affected by a few very expensive places).
print('Mean cost for two (Rs.):  ', round(df_india['Average Cost for two'].mean()))
print('Median cost for two (Rs.):', df_india['Average Cost for two'].median())
```

```python
# How does cost change by price range?
# This also shows what each price range MEANS in rupees.
df_india.groupby('Price range')['Average Cost for two'].agg(['count', 'min', 'mean', 'median', 'max']).round(0)
```

```python
# Which Indian cities are the most expensive?
# Again, only cities with at least MIN_RESTAURANTS (20) restaurants, so averages are reliable.
india_city_counts = df_india['City'].value_counts()
india_big_cities = india_city_counts[india_city_counts >=min_restaurant_count].index

india_city_cost = (
    df_india[df_india['City'].isin(india_big_cities)]
    .groupby('City')['Average Cost for two']
    .agg(['count', 'mean', 'median'])
    .round(0)
    .sort_values('mean', ascending=False)
)
india_city_cost.head(10)
```

```python
# Chart the 10 most expensive Indian cities by average cost
india_city_cost['mean'].head(10)[::-1].plot(kind='barh')
plt.title('Average cost for two (Rs.), top 10 Indian cities')
plt.xlabel('Rs.')
plt.show()
```
